#**3.3. Transformaciones y resumen estadístico**

## Transformación 1: Creación de rangos de descuento

Con el objetivo de facilitar el análisis de la relación entre descuentos y rentabilidad, se creó una nueva variable denominada **Rango_Descuento**.

La variable original **Discount** contiene valores numéricos continuos, lo que dificulta la comparación entre grupos. Por este motivo, los descuentos fueron agrupados en categorías: *Sin descuento*, *Bajo*, *Medio*, *Alto* y *Muy alto*.

Esta transformación permite comparar de manera más clara indicadores como ventas, ganancias y margen de rentabilidad entre distintos niveles de descuento, facilitando la identificación de posibles umbrales a partir de los cuales los descuentos dejan de ser rentables.

Posteriormente, esta nueva variable será utilizada para construir tablas resumen, gráficos comparativos y analizar si existe una relación entre el nivel de descuento aplicado y la rentabilidad obtenida.


```python
# Definición de los intervalos de descuento
bins = [-0.01, 0, 0.10, 0.20, 0.40, 1.00]

# Etiquetas para cada rango
labels = [
    "Sin descuento",
    "Bajo (0%-10%)",
    "Medio (10%-20%)",
    "Alto (20%-40%)",
    "Muy alto (>40%)"
]

# Creación de la nueva variable categórica
df["Rango_Descuento"] = pd.cut(
    df["Discount"],
    bins=bins,
    labels=labels
)

# Verificación de la transformación
print(df[["Discount", "Rango_Descuento"]].head())
print()
# Cantidad de registros por categoría
print(df["Rango_Descuento"].value_counts().sort_index())
```

## Transformación 2: Creación de una variable de margen de rentabilidad

Con el objetivo de evaluar el impacto de los descuentos sobre la rentabilidad, se creó una nueva variable denominada **Margen_Rentabilidad**.

La variable **Profit** indica la ganancia o pérdida obtenida en cada venta, pero no permite comparar directamente operaciones de diferente tamaño. Por ejemplo, una ganancia de 100 dólares puede ser muy buena en una venta de 500 dólares y poco significativa en una venta de 10.000 dólares.

Por este motivo, se calculó el margen de rentabilidad como la proporción de ganancia respecto de las ventas realizadas. Esta medida permite comparar de manera más adecuada la eficiencia económica de distintas operaciones, independientemente de su volumen de ventas.

La nueva variable será utilizada para analizar si los descuentos más elevados reducen la rentabilidad relativa de las ventas y para identificar posibles niveles de descuento asociados con márgenes bajos o negativos.


```python
# Margen de rentabilidad expresado como porcentaje
# Fórmula: ganancia / ventas

df["Margen_Rentabilidad_Porc"] = (
    df["Profit"] / df["Sales"]
) * 100

print(
    df[
        [
            "Sales",
            "Profit",
            "Margen_Rentabilidad_Porc"
        ]
    ].head()
)
print()
# Estadísticas descriptivas de la nueva variable
print(df["Margen_Rentabilidad_Porc"].describe())
```

## Transformación 3: Identificación de ventas con pérdida

Con el objetivo de analizar el impacto de los descuentos sobre los resultados económicos de cada operación, se creó una nueva variable denominada **Resultado_Venta**.

La variable **Profit** indica la ganancia o pérdida obtenida en cada transacción. Sin embargo, para ciertos análisis resulta más útil clasificar las operaciones según si fueron rentables o no, en lugar de observar únicamente el valor numérico de la ganancia.

Por este motivo, se creó una variable categórica que identifica cada venta como **Rentable** cuando la ganancia es positiva o nula, y **Con pérdida** cuando la ganancia es negativa.

Esta transformación permitirá calcular la proporción de operaciones con pérdidas en cada rango de descuento y evaluar si los descuentos más elevados aumentan la frecuencia de ventas no rentables.


```python
# Creación de una variable que clasifica cada venta
# según el resultado económico obtenido

df["Resultado_Venta"] = df["Profit"].apply(
    lambda x: "Con pérdida" if x < 0 else "Rentable"
)

# Verificación de la nueva variable
print(df[["Profit", "Resultado_Venta"]].head())
print()
# Cantidad de operaciones en cada categoría
print(df["Resultado_Venta"].value_counts())
```

## Estadístico de resumen 1: Profit promedio por categoría y rango de descuento

Con el objetivo de analizar si el impacto de los descuentos varía según el tipo de producto, se calculó la ganancia promedio (**Profit**) para cada combinación de categoría de producto y rango de descuento.

Mientras que el análisis general permite observar la relación entre descuentos y rentabilidad en todo el dataset, este agrupamiento permite identificar si algunas categorías son más sensibles a los descuentos que otras.

El cálculo de la ganancia promedio por grupo facilita la comparación entre categorías y permite detectar situaciones donde determinados niveles de descuento podrían generar una reducción significativa de la rentabilidad o incluso pérdidas.

Este análisis resulta relevante para la pregunta de investigación, ya que ayuda a determinar si existe un nivel de descuento que afecta negativamente la rentabilidad y si dicho efecto se presenta de manera uniforme en todas las categorías.


```python
# Profit promedio por categoría y rango de descuento

profit_promedio_categoria = (
    df.groupby(
        ["Category", "Rango_Descuento"]
    )["Profit"]
    .mean()
    .round(2)
    .reset_index()
)

# Mostrar resultados
print(profit_promedio_categoria)

# Ordenar para facilitar la lectura
profit_promedio_categoria = profit_promedio_categoria.sort_values(
    by=["Category", "Rango_Descuento"]
)

profit_promedio_categoria.head(15)
```

**Interpretación**: Con esta tabla podemos observar la ganancia promedio de cada sector, para saber si se repite el mismo patron con respecto a los distintos rangos de descuento y la ganancia.

Podemos observar que:

*   La categoria de **tecnologia** es la que tiene más variación de ganancia.
*   En las tres categorias hay una perdida de mas del 50% en descuentos muy alto

## Estadístico de resumen 2: Profit promedio por rango de descuento

Con el objetivo de analizar la relación entre descuentos y rentabilidad, se calculó la ganancia promedio (**Profit**) para cada rango de descuento definido previamente.

La utilización del profit promedio permite observar cómo varía el resultado económico de las operaciones a medida que aumenta el nivel de descuento aplicado. De esta manera, es posible evaluar si los descuentos contribuyen a mantener la rentabilidad o si, por el contrario, generan una reducción progresiva de las ganancias.

Este análisis constituye uno de los principales indicadores para responder la pregunta de investigación, ya que permite identificar si existe un nivel de descuento a partir del cual las promociones dejan de ser económicamente convenientes para la empresa.

Los resultados obtenidos servirán posteriormente como base para la elaboración de gráficos y para la interpretación de la relación entre descuentos y rentabilidad.


```python
# Profit promedio por rango de descuento

profit_promedio_descuento = (
    df.groupby("Rango_Descuento")["Profit"]
    .mean()
    .round(2)
    .reset_index()
)

# Mostrar tabla de resultados
print(profit_promedio_descuento)

# Mostrar ordenado según los rangos definidos
profit_promedio_descuento
```

**Interpretación**: El cálculo del profit promedio por rango de descuento permite analizar cómo varía la ganancia obtenida en las operaciones según el nivel de descuento aplicado.

Los resultados muestran una disminución progresiva del profit promedio a medida que aumentan los descuentos, ello sugeriría que las promociones reducen la rentabilidad de las ventas.

## Estadístico de resumen 3: Margen promedio por rango de descuento

Con el objetivo de evaluar cómo afecta el nivel de descuento a la rentabilidad de las operaciones, se calculó el margen de rentabilidad promedio para cada rango de descuento definido previamente.

A diferencia del profit promedio, que muestra la ganancia obtenida en términos absolutos, el margen de rentabilidad permite analizar qué proporción de las ventas se transforma efectivamente en ganancia. Esto facilita la comparación entre operaciones de distinto tamaño y proporciona una medida más precisa de la eficiencia económica de cada grupo.

El análisis del margen promedio resulta especialmente relevante para esta investigación, ya que permite determinar si el aumento de los descuentos reduce la rentabilidad relativa de las ventas y si existe un nivel de descuento a partir del cual dicha rentabilidad comienza a deteriorarse de forma significativa.

Los resultados obtenidos servirán para complementar el análisis de ganancias y aportar evidencia adicional sobre la relación entre descuentos y rentabilidad.


```python
# Resumen estadístico del margen por rango de descuento

resumen_margen = (
    df.groupby("Rango_Descuento")["Margen_Rentabilidad_Porc"]
    .agg(
        Margen_Promedio="mean",
        Margen_Minimo="min",
        Margen_Maximo="max",
        Cantidad_Operaciones="count"
    )
    .round(2)
)

resumen_margen
```

**Interpretación:** A medida que aumenta el descuento aplicado, el margen de rentabilidad promedio disminuye de forma sostenida. Los descuentos muy altos presentan márgenes negativos, lo que sugiere que el incremento en las ventas no compensa la reducción de ganancias provocada por la política de descuentos.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# =====================================================================
# CONFIGURACIÓN Y PREPARACIÓN DE VARIABLES LOCALES (ESCUDO SEGURO)
# =====================================================================
limites = [-0.01, 0.00, 0.10, 0.20, 0.40, 1.00]
nombres_rangos = ["Sin descuento", "Bajo (0%-10%)", "Medio (10%-20%)", "Alto (20%-40%)", "Muy alto (>40%)"]
df['Rango_descuento_graficos'] = pd.cut(df['Discount'], bins=limites, labels=nombres_rangos)

# Tratamiento contra ZeroDivisionError: evita fallos matemáticos si Sales == 0
df['Margen_seguro_graficos'] = np.where(df['Sales'] > 0, (df['Profit'] / df['Sales']) * 100, 0)


# =====================================================================
# Gráfico 1: Histograma de distribución de descuentos (3.4.a)
# =====================================================================
plt.figure(figsize=(9, 5))
sns.histplot(data=df, x="Discount", bins=15, kde=True, color="steelblue")

plt.title("Distribución General de los Descuentos Aplicados", fontsize=12, fontweight='bold', pad=12)
plt.xlabel("Porcentaje de Descuento (0.0 = 0% a 1.0 = 100%)", fontsize=10)
plt.ylabel("Cantidad de Operaciones", fontsize=10)
plt.grid(axis='y', linestyle=':', alpha=0.6)

plt.savefig("grafico_1_distribucion.png", bbox_inches='tight', dpi=300)
plt.show()


# =====================================================================
# Gráfico 2: Barras de ganancia promedio (3.4.b)
# =====================================================================
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x="Rango_descuento_graficos", y="Profit", order=nombres_rangos, palette="muted", errorbar=None)

plt.title("Ganancia Promedio (USD) por Nivel de Descuento", fontsize=12, fontweight='bold', pad=12)
plt.xlabel("Rangos de Descuento", fontsize=10)
plt.ylabel("Ganancia Promedio en USD", fontsize=10)
plt.axhline(0, color='black', linewidth=1, linestyle='-')
plt.grid(axis='y', linestyle=':', alpha=0.6)

plt.savefig("grafico_2_barras_profit.png", bbox_inches='tight', dpi=300)
plt.show()


# =====================================================================
# Gráfico 3: Dispersión y Margen de Rentabilidad (3.4.c)
# =====================================================================
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x="Discount", y="Margen_seguro_graficos", alpha=0.5, color="darkorange")

plt.title("Relación entre Porcentaje de Descuento y Margen de Rentabilidad", fontsize=12, fontweight='bold', pad=12)
plt.xlabel("Descuento (Proporción de 0.0 a 1.0)", fontsize=10)
plt.ylabel("Margen de Rentabilidad (%)", fontsize=10)
plt.axhline(y=0, color='red', linestyle='--', linewidth=1.5, alpha=0.8)
plt.grid(True, linestyle=':', alpha=0.5)

plt.savefig("grafico_3_dispersion_margen.png", bbox_inches='tight', dpi=300)
plt.show()
```
